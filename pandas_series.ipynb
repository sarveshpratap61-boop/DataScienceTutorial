**Introduction to Pandas**

Pandas is a popular, **open-source Python library used for data
manipulation, cleaning, and analysis**. Built on top of NumPy, it allows you to **work with structured, tabular data—much like a programmable Excel spreadsheet or SQL table**.

The name "pandas" comes from **panel data**(an econometrics term for multi-dimensional datasets) and Python data analysis.

# Pandas Series
In Python, Pandas Series is a one-dimensional labeled array that can store **data of any type, such as integers, strings, floating-point numbers**, etc.

A Series contains two main components:
1. **Index**: Labels assigned to each element.
2. **Values**: Actual data stored in the Series.

**Handling Duplicate Values**
In real-world datasets, duplicate records can create incorrect analysis. Pandas provides methods to identify and remove duplicate values.

We will learn:
1. duplicated()
2. drop_duplicates()
3. unique()
4. nunique()

```python
import pandas as pd
students = pd.Series([
    'Amit', 'Rahul', 'Priya',
    'Amit', 'Neha', 'Rahul'
])
print(students.duplicated()) #return True for unique values only
```

```python
import pandas as pd
students = pd.Series([
    'Amit', 'Rahul', 'Priya',
    'Amit', 'Neha', 'Rahul'
])
print(students.duplicated(keep='first')) #Keep first occurence as False and Second as True
```

```python
import pandas as pd
students = pd.Series([
    'Amit', 'Rahul', 'Priya',
    'Amit', 'Neha', 'Rahul'
])
print(students.duplicated(keep='last')) #Keeps the last occurrence as False and marks earlier occurrences as True.
```

```python
import pandas as pd
students = pd.Series([
    'Amit', 'Rahul', 'Priya',
    'Amit', 'Neha', 'Rahul'
])
x=students.drop_duplicates()
print(x)
```

```python
#reset_index after remove duplicates
import pandas as pd
students = pd.Series([
    'Amit', 'Rahul', 'Priya',
    'Amit', 'Neha', 'Rahul'
])
x=students.drop_duplicates().reset_index(drop=True)
print(x)
```

```python
#unique and nuunique
import pandas as pd
students = pd.Series([
    'Amit', 'Rahul', 'Priya',
    'Amit', 'Neha', 'Rahul','Amit',None
])
print(students.unique())
print("Number of ",students.nunique())
```

```python
#include missing values
import pandas as pd
students = pd.Series([
    'Amit', 'Rahul', 'Priya',
    'Amit', 'Neha', 'Rahul','Amit',None
])
print("Number of ",students.nunique(dropna=False))
```

**Pandas Series – Date and Time Operations (.dt)**

In Data Science, dates and time are frequently used in sales analysis, employee attendance, delivery tracking, banking transactions and business reporting.

Pandas provides the .dt accessor to extract and manipulate date and time components.

**1. What is .dt?**

Definition: .dt is a Pandas accessor used to perform date and time operations on a Series containing datetime values.

Syntax:**series.dt.attribute**

```python
import pandas as pd
dates=pd.Series(['2026-11-22','2026-10-19','2025-01-25'])
print(dates)
x=pd.to_datetime(dates)
print(x)
```

```python
import pandas as pd
dates=pd.Series(['2026-11-22','2026-10-19','2025-01-25'])
dates=pd.to_datetime(dates)
print("Year : ")
print(dates.dt.year)
print("Month : ")
print(dates.dt.month)
print("Days ")
print(dates.dt.day)
print("Month Name : ")
print(dates.dt.month_name())
print("Day Name : ")
print(dates.dt.day_name())
print("Quarter : ")
print(dates.dt.quarter)
print("Week Number")
print(dates.dt.dayofweek)
```

```python
import pandas as pd
dates=pd.Series(['2026-11-30','2026-10-19','2025-01-25'])
dates=pd.to_datetime(dates)
print("Year : ")
print(dates.dt.year)
print(dates.dt.is_month_end)
```

```python
#time
import pandas as pd
time_data = pd.Series([
    '2026-01-15 10:30:45',
    '2026-02-20 14:45:30',
    '2026-03-10 18:20:15'
])

time_data = pd.to_datetime(time_data)
print(time_data)
print(time_data.dt.hour)
print(time_data.dt.minute)
print(time_data.dt.second)
```
